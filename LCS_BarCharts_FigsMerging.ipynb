```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

files = [
    "results/tmo_results_pbmc_cmmd/lcs_comparison.png",
    "results/tmo_results_mouse_brain_cmmd/lcs_comparison.png",
    "results/tmo_results_human_brain_cmmd/lcs_comparison.png",
    "results/tmo_results_mouse_kidney_cmmd/lcs_comparison.png"
]
```

```python
titles = ["PBMC", "Mouse Brain", "Human Brain", "Mouse Kidney"]
letters = ["a", "b", "c", "d"]
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
for ax, f, t, l in zip(axes.flat, files, titles, letters):
    img = mpimg.imread(f)
    ax.imshow(img)
    ax.set_title(t, fontsize=14)
    ax.axis('off')
    # Add panel letter in the top‑left corner
    ax.text(0.02, 0.98, f"({l})", transform=ax.transAxes,
            fontsize=16, fontweight='bold', va='top', ha='left',
            color='black', backgroundcolor='white')

plt.tight_layout()
plt.savefig("fig4_lcs_comparison.pdf", dpi=300, bbox_inches='tight')
plt.savefig("fig4_lcs_comparison.png", dpi=300, bbox_inches='tight')
plt.show()
```
