```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

files = [
    "./results/tmo_results_pbmc_cmmd/biphasic_candidates_pbmc.png",
    "./results/tmo_results_mouse_brain_cmmd/biphasic_candidates_mouse_brain.png",
    "./results/tmo_results_human_brain_cmmd/biphasic_candidates_human_brain.png",
    "./results/tmo_results_mouse_kidney_cmmd/biphasic_candidates_mouse_kidney.png",
]
```

```python
titles = ["PBMC", "Mouse Brain", "Human Brain", "Mouse Kidney"]
letters = ["a", "b", "c", "d"]

fig, axes = plt.subplots(2, 2, figsize=(14, 12))
for ax, f, t, l in zip(axes.flat, files, titles, letters):
    img = mpimg.imread(f)
    ax.imshow(img)
    ax.set_title(t, fontsize=14)
    ax.axis('off')
    # Add panel letter in the top‑left corner
    ax.text(0.02, 0.98, f"({l})", transform=ax.transAxes,
            fontsize=16, fontweight='bold', va='top', ha='left',
            color='black', backgroundcolor='white')

plt.tight_layout()
plt.savefig("S6_biphasic_profiles_all.pdf", dpi=300, bbox_inches='tight')
plt.savefig("S6_biphasic_profiles_all.png", dpi=300, bbox_inches='tight')
plt.show()
```
