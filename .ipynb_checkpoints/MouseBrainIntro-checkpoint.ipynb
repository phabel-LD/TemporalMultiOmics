```python
import anndata as ad
import scanpy as sc
import numpy as np
from scipy.sparse import csr_matrix
```

```python
# RNA AnnData
rna_path = r"C:\Users\phabe\OneDrive\Escritorio\YearProject\PROJECT\DATA\MouseBrain_GEX_Def.h5ad"
adata_rna = ad.read_h5ad(rna_path)

# ATAC AnnData – correct filename
atac_path = r"C:\Users\phabe\OneDrive\Escritorio\YearProject\PROJECT\DATA\MouseBrain_Peak_Def.h5ad"
adata_atac = ad.read_h5ad(atac_path)

# Make variable names unique (suppresses warning)
adata_rna.var_names_make_unique()
adata_atac.var_names_make_unique()

print("RNA shape:", adata_rna.shape)
print("ATAC shape:", adata_atac.shape)
```

```python
common_cells = adata_rna.obs_names.intersection(adata_atac.obs_names)
print(f"Common cells: {len(common_cells)}")

# Subset both to common cells
adata_rna = adata_rna[common_cells, :]
adata_atac = adata_atac[common_cells, :]

# Ensure same order (ATAC in same order as RNA)
adata_atac = adata_atac[adata_rna.obs_names, :]

print("After alignment:")
print("RNA:", adata_rna.shape)
print("ATAC:", adata_atac.shape)
```

```python
adata_combined = adata_rna.copy()
# Store ATAC counts (sparse) in .obsm
adata_combined.obsm['ATAC'] = adata_atac.X.copy()

print("Combined object:")
print("RNA in .X:", adata_combined.X.shape)
print("ATAC in .obsm['ATAC']:", adata_combined.obsm['ATAC'].shape)
```

```python
# Cell 5 (corrected for scanpy version)
if 'pseudotime' not in adata_combined.obs:
    print("Computing pseudotime with DPT...")
    sc.pp.pca(adata_combined, n_comps=30)
    sc.pp.neighbors(adata_combined)
    sc.tl.diffmap(adata_combined)
    
    # Set root cell in uns
    root_idx = adata_combined.obsm['X_diffmap'][:, 0].argmin()
    adata_combined.uns['iroot'] = root_idx
    print(f"Using root cell index: {root_idx}")
    
    # Run DPT (without root_idx argument; uses uns['iroot'])
    sc.tl.dpt(adata_combined, n_branchings=0)
    
    # Check if dpt_pseudotime was created
    if 'dpt_pseudotime' in adata_combined.obs:
        pt = adata_combined.obs['dpt_pseudotime'].values
        pt = (pt - pt.min()) / (pt.max() - pt.min())
        adata_combined.obs['pseudotime'] = pt
    else:
        # Fallback: use first diffusion component
        print("Warning: dpt_pseudotime not found. Using first diffusion component instead.")
        pt = adata_combined.obsm['X_diffmap'][:, 0]
        pt = (pt - pt.min()) / (pt.max() - pt.min())
        adata_combined.obs['pseudotime'] = pt

print("Pseudotime range:", adata_combined.obs['pseudotime'].min(), "-", adata_combined.obs['pseudotime'].max())
```

```python
output_path = "mouse_brain_combined.h5ad"
adata_combined.write(output_path)
print(f"Saved to {output_path}")
```
