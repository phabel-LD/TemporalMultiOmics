```python
# # 1. Import libraries and set up paths

import tmopy as tmo
import scanpy as sc
import matplotlib.pyplot as plt
import warnings
import random
import os
random.seed(8)
warnings.filterwarnings("ignore")
```

```python
# Paths to raw data (adjust if necessary)
h5_file = "./data/10x_mouse_brain/e18_mouse_brain_fresh_5k_filtered_feature_bc_matrix.h5"
atac_annotation_file = "./data/10x_mouse_brain/e18_mouse_brain_fresh_5k_atac_peak_annotation.tsv"
output_dir = "./results/tmo_results_mouse_brain"
os.makedirs(output_dir, exist_ok=True)
```

```python
# # 2. Load and preprocess data

adata = tmo.pp.read_10x_multiome(h5_file, atac_annotation_file)
print(adata)
```

```python
# Compute gene-level ATAC matrix (peaks -> genes)
adata = tmo.pp.compute_gene_level_atac(adata)
print("Gene-level ATAC matrix shape:", adata.obsm['ATAC_gene'].shape)
```

```python
# Compute pseudotime, using Pax6 as early marker to orient the trajectory
adata = tmo.pp.compute_pseudotime(adata, root_marker="Pax6")
print(f"Pseudotime range: {adata.obs['pseudotime'].min():.3f} – {adata.obs['pseudotime'].max():.3f}")
print(f"Dataset reversed: {adata.uns.get('pseudotime_reversed', False)}")
```

```python
# Optional: filter to highly variable genes and top peaks (speeds up training) -> But it may reduce stat power
#adata = tmo.filter_genes_and_peaks(adata, n_top_genes=2000, n_top_peaks=20000)
#print(f"After filtering: {adata.shape}")
```

```python
# # 3. Annotate RNA components with GO terms (takes a few minutes)

component_annotation = tmo.tl.annotate_components(
    adata, n_components=50, n_top_genes=100, n_top_go=5, delay=2.0, organism='mouse'
)
component_annotation.head()
```

```python
# # 4. Train asymmetric TMO model

model = tmo.tl.train_asymmetric(
    adata,
    epochs=50,
    val_interval=5,
    lambda_lag=0.1,
    batch_size=32,
    lr=1e-3,
    device='cpu',          # change to 'cuda' if GPU available
    save_model=os.path.join(output_dir, "tmo_asymmetric_best.pt"),
    output_dir=output_dir
)
```

```python
# # 5. Evaluate the model and generate Δτ heatmap (Figure 1)

delta_tau, pt_grid, lcs = tmo.tl.evaluate_asymmetric(
    adata,
    model_path=os.path.join(output_dir, "tmo_asymmetric_best.pt"),
    n_components=50
)
print(f"Lag Concordance Score (LCS): {lcs:.4f}")
```

```python
# Prepare row labels: use component names from annotation (or generic)
row_labels = component_annotation['Component'].tolist()[:50] # first 50 components
# Plot heatmap
fig1 = tmo.pl.plot_delta_tau(
    delta_tau,
    pt_grid,
    row_labels,
    title="Mouse brain Δτ heatmap",
    save=os.path.join(output_dir, "fig1_delta_tau.pdf"),
    show=False
)
plt.close(fig1)
```

```python
# ## 6. Generate ATAC‑RNA correlation heatmap (Figure 2)

fig2 = tmo.pl.plot_correlation(
    adata,
    n_bins=10,
    save=os.path.join(output_dir, "fig2_correlation.pdf"),
    show=False
)
plt.close(fig2)
```

```python
# ## 7. Plot training curves (Figure 3)
# The training script saved a CSV file in the output directory. If you did not change the default output directory, it is in ./tmo_results/training_metrics.csv.

metrics_csv = os.path.join(output_dir, "training_metrics.csv")
if os.path.exists(metrics_csv):
    fig3 = tmo.pl.plot_training_curves(metrics_csv, save=os.path.join(output_dir, "fig3_training_curves.pdf"), show=False)
    plt.close(fig3)
else:
    print("Training metrics CSV not found; please adjust path.")
```

```python
# ## 8. (Optional) ChIP‑seq validation using Pax6 target genes (mouse brain)
# For mouse brain, we use a pre‑computed list of Pax6 target genes from ChIP‑Atlas.
# The file 'Pax6_top500_genes.txt' contains the top 500 genes with highest binding score.

target_file = "./data/10x_mouse_brain/Pax6.5_top500_genes.txt"
if os.path.exists(target_file):
    print("Running ChIP‑seq validation for Pax6...")
    chip_res = tmo.tl.validate_chipseq(
        adata,
        model_path=os.path.join(output_dir, "tmo_asymmetric_best.pt"),
        target_genes_file=target_file,
        match_expression=True,
        n_components=50,
        alternative='two-sided',
    )
    print(f"ChIP‑seq validation p‑value: {chip_res['p_value']:.4e}")
    print(f"Target Δτ mean: {chip_res['mean_target']:.6f}, background Δτ mean: {chip_res['mean_background']:.6f}")
else:
    print(f"Target gene file {target_file} not found. Skipping ChIP‑seq validation.")
```

```python
if(chip_res['p_value']):
    # Plot violin plot
    p_val = chip_res['p_value']
    fig4 = tmo.pl.plot_violin_validation(
        chip_res['target_lags'], chip_res['background_lags'],
        target_label="Pax6 targets",
        background_label="Background",
        title=f"Pax6 ChIP‑seq validation (p = {p_val:.2e}) (mouse brain)",
        ylabel="Δτ (pseudotime units)",
        save=os.path.join(output_dir, "fig4_chipseq_pax6.pdf"),
        show=False
    )
    plt.close(fig4)
    print("ChIP‑seq validation figure saved to fig4_chipseq_pax6.pdf")
else:
    print("Run ChIP-seq validation first, to compute p-values.")
```

```python
# ## 9. Summary
print("\nAnalysis completed. Results saved in:", output_dir)
print("Figure 1: Δτ heatmap (fig1_delta_tau.pdf)")
print("Figure 2: ATAC‑RNA correlation (fig2_correlation.pdf)")
print("Figure 3: Training curves (fig3_training_curves.pdf)")
```
